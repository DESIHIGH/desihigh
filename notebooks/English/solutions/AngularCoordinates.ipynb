## 1 - Map Projections

When we make maps of our planet Earth, we are faced with the challenge of representing a three-dimensional globe on a two-dimensional map. To overcome this challenge, we use a **map projection** of the Earth's surface. A projection is a mathematical formula that we can use to transfer every point on the surface of the Earth onto a flat map. Below are some examples of different map projections next to a photo of the Earth.

We will use the **Mollweide projection** for the maps we make in this notebook. This projection preserves the physical sizes of the continents, but to do so, it distorts their shapes. Different types of projections preserve different properties of the globe, such as areas, angles, or distances.

## 2 - Longitude and Latitude 

To tell where we are located on our planet Earth, we can use **longitude (long.)** and **latitude (lat.)** coordinates. Longitude lines are vertical lines on the Earth's surface that start and end at the Earth's north and south poles. Latitude lines are horizontal lines on the Earth's surface that are parallel to the equator. As we will see, **longitude and latitude lines are  angles that have units of degrees.**

## 3 - Right Ascension and Declination 

Before we introduce a new coordinate system, let's see what a map of the night sky looks like. Just like with maps of the Earth, maps of the night sky use projections to depict the three-dimensional night sky on a two-dimensional surface. **Run the code below to load in and plot a map of the night sky.**

```python
# load in night sky map
sky_map = hp.read_map('../../../data/wmap_band_iqumap_r9_7yr_W_v4.fits')

# plot a map of the night sky
projections.plot_sky_map(sky_map, 0, 0)
```

Most of this map is dark, but the curving white band that runs across the map is the band of the Milky Way galaxy. If you are ever in a dark countryside on a cloudless night, you can see this band of stars and gas in the sky above you.

When making maps of the night sky, we use angular coordinates called **right ascension (RA)** and **declination (dec)**. These coordinates have units of degrees and work in a similar way to longitude and latitude. Any point in the sky sits on a line of right ascension and a line of declination. For example, the red star in the plot you made above is sitting on a blue line of right ascension and a pink line of declination. 

Let's try changing the location of the red star. **Run the code below,** which lets us set the star's location using the `right ascension` and `declination` variables.

```python
# set the right ascension and declination for the starred location on the map
right_ascension = 0
declination = 0

# plot a map of the night sky
projections.plot_sky_map(sky_map, right_ascension, declination)
```

Now we are going to fix `right ascension` and `declination` to `0`, and we are going to introduce a new variable named `right_ascension_offset`. This variable works similarly to the `longitude_offset` variable that we saw earlier. **Run the code below to set `right_ascension_offset` to `0`.**


```python
# set a longitude offset for the map
right_ascension_offset = 0

# plot a map of the night sky
projections.plot_transformed_sky_map(sky_map, 0, 0, right_ascension_offset, 0, 0)
```
